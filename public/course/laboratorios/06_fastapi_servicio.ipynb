```python
from fastapi import FastAPI
from pydantic import BaseModel, Field
from fastapi.testclient import TestClient

app=FastAPI(title="Servicio de decisión NEXUS")
class Solicitud(BaseModel):
 probabilidad: float=Field(ge=0,le=1)
 consentimiento: bool

@app.get("/health")
def health(): return {"status":"ok"}

@app.post("/decide")
def decide(s:Solicitud):
 if not s.consentimiento:
  return {"accion":"no_actuar","explicacion":"Sin consentimiento","revision_humana":False}
 if .45 <= s.probabilidad <= .55:
  return {"accion":"abstener","explicacion":"Evidencia insuficiente","revision_humana":True}
 return {"accion":"intervenir" if s.probabilidad>.55 else "esperar","explicacion":"Política de umbral 0.55","revision_humana":False}

client=TestClient(app)
print(client.get("/health").json())
print(client.post("/decide",json={"probabilidad":.8,"consentimiento":True}).json())
```
