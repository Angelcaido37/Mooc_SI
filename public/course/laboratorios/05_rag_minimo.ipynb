## A. Flujo visual (20 min)

Ordena estos bloques: documentos → fragmentos → consulta → similitud → fragmento recuperado → respuesta con fuente. Indica dónde puede aparecer una respuesta sin respaldo.


```python
import re, math
docs={"D1":"BFS explora por niveles y usa una cola.","D2":"A estrella combina costo acumulado y una heurística.","D3":"Una regla fija automatiza una decisión sin aprender."}
def tokens(t): return set(re.findall(r"[a-záéíóúñ]+",t.lower()))
def recuperar(pregunta):
 q=tokens(pregunta); ranking=[]
 for clave,texto in docs.items():
  d=tokens(texto); score=len(q&d)/math.sqrt(max(1,len(q)*len(d)))
  ranking.append((score,clave,texto))
 return max(ranking)
print(recuperar("¿Qué algoritmo usa una heurística?"))
```

## TU TURNO (50 min)

Agrega cinco fragmentos de tus propias notas. Si la similitud es menor a 0.15, el sistema debe responder «evidencia insuficiente». Devuelve siempre el identificador de fuente.


```python
def responder(pregunta):
 score,fuente,fragmento=recuperar(pregunta)
 # Completa la abstención y una respuesta sustentada
 return {"score":score,"fuente":fuente,"evidencia":fragmento}
print(responder("¿Cuándo se usa una cola?"))
```
