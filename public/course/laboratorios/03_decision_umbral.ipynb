## A. Exploración (20 min)

En una hoja de cálculo crea un control de umbral entre 0 y 1. Antes de moverlo, define cuál error cuesta más: intervenir sin necesidad o no intervenir a tiempo.


```python
datos=[(0.92,1),(0.76,1),(0.63,0),(0.51,1),(0.40,0),(0.22,0)]
def evaluar(umbral):
 pred=[int(p>=umbral) for p,_ in datos]; real=[y for _,y in datos]
 tp=sum(a==b==1 for a,b in zip(pred,real)); fp=sum(a==1 and b==0 for a,b in zip(pred,real))
 fn=sum(a==0 and b==1 for a,b in zip(pred,real)); tn=sum(a==b==0 for a,b in zip(pred,real))
 return {"TP":tp,"FP":fp,"FN":fn,"TN":tn}
for u in (.4,.6,.8): print(u,evaluar(u))
```
