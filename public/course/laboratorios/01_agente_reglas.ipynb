## A. Antes de programar (15 min)

Dibuja con bloques: **entrada → regla → acción → explicación**. Usa tres entradas ficticias: días para vencer, número de recordatorios y solicitud de no recibir mensajes.


```python
casos = [
 {"dias": 3, "recordatorios": 0, "consentimiento": True},
 {"dias": -1, "recordatorios": 1, "consentimiento": True},
 {"dias": 2, "recordatorios": 0, "consentimiento": False}
]

def decidir(caso):
    if not caso["consentimiento"]:
        return {"accion":"no_enviar", "explicacion":"No existe consentimiento"}
    if caso["dias"] < 0:
        return {"accion":"revisar", "explicacion":"La devolución ya venció"}
    if caso["dias"] <= 3 and caso["recordatorios"] == 0:
        return {"accion":"recordar", "explicacion":"Vence en tres días o menos"}
    return {"accion":"esperar", "explicacion":"No se activa ninguna regla"}

for c in casos: print(c, decidir(c))
```

Agrega una regla de abstención cuando falte un dato. Después escribe cuatro pruebas: caso normal, límite, dato faltante y consentimiento falso.
